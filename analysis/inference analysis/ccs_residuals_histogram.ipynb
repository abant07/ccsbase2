# CCS Residual Histogram

This notebook loads `pretrained/testset_predictions.csv`, computes residuals (`CCS_True - CCS_Pred`), and plots a histogram.

```python
# Global style settings
plt.rcParams.update({
    "font.family": "Arial",
    "font.sans-serif": ["Arial", "Helvetica", "DejaVu Sans"], # Standard crisp fonts
    "axes.linewidth": 3,               # Thicker axis lines
    "axes.spines.top": True,            # Remove top spine
    "axes.spines.right": True,          # Remove right spine
    "xtick.major.width": 3,            # Match tick thickness to axis
    "xtick.major.size": 8,
    "ytick.major.size": 8,
    "ytick.major.width": 3,
    "xtick.direction": "out",            # Ticks point outside
    "ytick.direction": "out",
    "font.size": 14,
    "axes.labelsize": 18,
    "axes.labelweight": "normal",
})
```

```python
import pandas as pd
import matplotlib.pyplot as plt

csv_path = "/Users/ryannguyen/Desktop/ccsbase2/pretrained/testset_predictions.csv"

df = pd.read_csv(csv_path)

# Residuals: true minus predicted
residuals = df["CCS_True"] - df["CCS_Pred"]

plt.figure(figsize=(6, 5))
plt.hist(residuals, bins=50, color="#4C72B0", edgecolor="white")
# plt.title("Residuals (CCS_True - CCS_Pred)")
plt.xlabel("Residual ($\mathrm{\AA^2}$)")
plt.ylabel("Frequency")
plt.xlim(-20, 20)
plt.ylim(0, 2000)
plt.tight_layout()

plt.savefig(
    "Residual_CCS.png",
    dpi=600,
    bbox_inches="tight"
)
plt.show()
```
